```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
palette = ["#4A90E2", "#E74C3C"]

# Load data and remove IDs
df = pd.read_csv("Churn_Modelling.csv")
df_clean = df.drop(
    columns=["RowNumber", "CustomerId", "Surname"], errors="ignore"
)

# 1. Target Distribution
fig, ax = plt.subplots(figsize=(5, 3.5))
churn_pct = df_clean["Exited"].value_counts(normalize=True) * 100
bars = ax.bar(
    ["Retained (0)", "Exited (1)"],
    churn_pct,
    color=palette,
    width=0.45,
    edgecolor="black",
)
ax.set_ylabel("Percentage (%)", fontsize=9)
ax.set_title("Customer Churn Distribution", fontsize=11, weight="bold")
ax.set_ylim(0, 100)
for bar in bars:
  h = bar.get_height()
  ax.text(
      bar.get_x() + bar.get_width() / 2,
      h + 2,
      f"{h:.1f}%",
      ha="center",
      va="bottom",
      fontsize=9,
      weight="bold",
  )
plt.tight_layout()
plt.savefig("fig1_churn_distribution.png", dpi=90)
plt.close()

# 2. Age Distribution
fig, ax = plt.subplots(figsize=(6, 3.5))
sns.histplot(
    data=df_clean,
    x="Age",
    hue="Exited",
    kde=True,
    bins=25,
    palette=palette,
    ax=ax,
    element="step",
)
ax.set_title("Age Distribution by Churn Status", fontsize=11, weight="bold")
ax.set_xlabel("Customer Age (Years)", fontsize=9)
ax.set_ylabel("Count", fontsize=9)
ax.legend(title="Status", labels=["Exited (1)", "Retained (0)"], fontsize=8)
plt.tight_layout()
plt.savefig("fig2_age_churn_distribution.png", dpi=90)
plt.close()

# 3. Geography & Gender Churn Rates
fig, axes = plt.subplots(1, 2, figsize=(8, 3.5))
geo_churn = df_clean.groupby("Geography")["Exited"].mean().reset_index()
sns.barplot(
    data=geo_churn,
    x="Geography",
    y="Exited",
    ax=axes[0],
    palette="Blues_r",
    edgecolor="black",
)
axes[0].set_title("Churn Rate by Country", fontsize=10, weight="bold")
axes[0].set_ylabel("Churn Rate", fontsize=9)
axes[0].set_ylim(0, 0.45)
for p in axes[0].patches:
  axes[0].annotate(
      f"{p.get_height()*100:.1f}%",
      (p.get_x() + p.get_width() / 2.0, p.get_height() + 0.01),
      ha="center",
      weight="bold",
      fontsize=8,
  )

gender_churn = df_clean.groupby("Gender")["Exited"].mean().reset_index()
sns.barplot(
    data=gender_churn,
    x="Gender",
    y="Exited",
    ax=axes[1],
    palette="Purples_r",
    edgecolor="black",
)
axes[1].set_title("Churn Rate by Gender", fontsize=10, weight="bold")
axes[1].set_ylabel("Churn Rate", fontsize=9)
axes[1].set_ylim(0, 0.45)
for p in axes[1].patches:
  axes[1].annotate(
      f"{p.get_height()*100:.1f}%",
      (p.get_x() + p.get_width() / 2.0, p.get_height() + 0.01),
      ha="center",
      weight="bold",
      fontsize=8,
  )
plt.tight_layout()
plt.savefig("fig3_categorical_churn_rates.png", dpi=90)
plt.close()

# 4. Balance vs NumOfProducts
fig, ax = plt.subplots(figsize=(6, 3.5))
sns.boxplot(
    data=df_clean,
    x="NumOfProducts",
    y="Balance",
    hue="Exited",
    palette=palette,
    ax=ax,
    fliersize=2,
)
ax.set_title("Balance vs. Products Held", fontsize=11, weight="bold")
ax.set_xlabel("Number of Products", fontsize=9)
ax.set_ylabel("Balance ($)", fontsize=9)
ax.legend(title="Status", labels=["Retained (0)", "Exited (1)"], fontsize=8)
plt.tight_layout()
plt.savefig("fig4_balance_products_interaction.png", dpi=90)
plt.close()

print("All 4 figures generated successfully!")
```
